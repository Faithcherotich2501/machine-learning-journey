```python
import matplotlib.pyplot as plt
import seaborn as sns

# Get the count of each sex
sex_counts = df['Sex'].value_counts()

# Plot the counts of each sex
plt.figure(figsize=(8, 5))
sns.barplot(x=sex_counts.index, y=sex_counts.values)
plt.title('Count of Passengers by Sex')
plt.xlabel('Sex')
plt.ylabel('Count')
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Group by Pclass and Survived and unstack to get counts
pclass_survival = df.groupby('Pclass')['Survived'].value_counts().unstack().fillna(0)

# Calculate survival rate
pclass_survival['Survival Rate'] = pclass_survival[1] / (pclass_survival[0] + pclass_survival[1])

# Plot the survival rate by Pclass
plt.figure(figsize=(8, 5))
sns.barplot(x=pclass_survival.index, y='Survival Rate', data=pclass_survival)
plt.title('Survival Rate by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Survival Rate')
plt.xticks(ticks=[0, 1, 2], labels=['1st Class', '2nd Class', '3rd Class'])
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Group by Sex and Survived and unstack to get counts
sex_survival = df.groupby('Sex')['Survived'].value_counts().unstack().fillna(0)

# Plot the survival counts by Sex
sex_survival.plot(kind='bar', stacked=True, figsize=(8, 5))
plt.title('Survival Count by Sex')
plt.xlabel('Sex')
plt.ylabel('Count')
plt.xticks(rotation=0)
plt.legend(['Did Not Survive', 'Survived'])
plt.show()
```
